<h2 style="color:#90EE90; text-align:left;">
Handling Outliers
</h2>

<p style="color:#90EE90; line-height:1.8;">
Outliers are extreme values that can affect analysis results. In this project, 
outliers in cost-related columns are identified and handled to ensure more accurate insights.
</p>

</div>

```python
df['ITEM TOTAL COST'] = pd.to_numeric(df['ITEM TOTAL COST'], errors='coerce')
```

```python
Q1 = df['ITEM TOTAL COST'].quantile(0.25)
Q3 = df['ITEM TOTAL COST'].quantile(0.75)

IQR = Q3 - Q1

df = df[(df['ITEM TOTAL COST'] >= Q1 - 1.5*IQR) & 
        (df['ITEM TOTAL COST'] <= Q3 + 1.5*IQR)]
```

```python
df.shape
```

```python
print(Q1,Q3)
```

<h3 style="color:#AF601A;">Outlier Handling Insight</h3>

<p>
Outliers were identified and removed using the IQR method on the 'ITEM TOTAL COST' column. 
The 'TOTAL AMOUNT' column was not suitable for outlier detection due to a large number of zero values. 
After removing outliers, the dataset became more consistent and suitable for analysis.
</p>

<h3 style="color:#90EE90; text-align:left;">
Creating Expense Categories
</h3>

<p style="color:#90EE90; line-height:1.8;">
To simplify analysis, detailed expense types are grouped into broader categories 
such as repair & maintenance, supplies, equipment, and services. 
This helps in better understanding overall spending patterns.
</p>

</div>

```python
def categorize_expense(x):
    if 'repair' in x:
        return 'repair & maintenance'
    elif 'supplies' in x:
        return 'supplies'
    elif 'equipment' in x:
        return 'equipment'
    elif 'services' in x:
        return 'services'
    elif 'software' in x or 'hardware' in x:
        return 'it related'
    elif 'vehicle' in x:
        return 'vehicle'
    else:
        return 'other'

df['EXPENSE CATEGORY'] = df['EXPENSE TYPE DESCRIPTION'].apply(categorize_expense)
```
